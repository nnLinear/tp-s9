```python
import numpy as np

X=np.array([
    [0 , 0],
    [0 , 1],
    [1 , 0],
    [1 , 1],
]).T

y=np.array([
    [0],
    [1],
    [1],
    [0]
]).T

print(X.shape , y.shape)
```

```python
def initialisation(ts):
    ans = []
    for i in ts:
      
        ans.append((np.random.randn(*i) *0.1 ,np.zeros((i[0] , 1))))

    return ans

params =initialisation([(3 , 2) , (1 , 3) ])

print(len(params))
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python

def forward(X , parametres):
    ans = []
    A=X
    for params in parametres:
        Z =params[0] @ A  + params[1]
        A = sigmoid(Z)
        ans.append((Z , A))

    return ans , ans[len(params) - 1][1]


his , y_hat = forward(X , params)
print(y_hat)
```

```python
def mse_error(Y , Y_hat):
    return 1 / (2 * Y.shape[0]) * np.sum((Y_hat - Y)**2)

print(mse_error(y , y_hat[0]))
```

```python
def backward(X, Y, Y_hat, cache, parameters):
    m = Y.shape[1]                                   # 4 observations
    L = len(parameters)
    grads = [None] * L

    dZ = (Y_hat - Y) * Y_hat * (1 - Y_hat)           # delta[2], (1,4)

    for l in range(L - 1, -1, -1):
        A_prev = X if l == 0 else cache[l - 1][1]

        dW = dZ @ A_prev.T / m
        db = np.sum(dZ, axis=1, keepdims=True) / m
        grads[l] = (dW, db)

        if l > 0:
            W = parameters[l][0]
            A = cache[l - 1][1]
            dZ = (W.T @ dZ) * A * (1 - A)

    return grads

grads = backward(X, y, y_hat, his, params)    # no .T here
for dW, db in grads:
    print(dW.shape, db.shape)
```
