```python
import numpy as np

np.random.seed(42)
def initialisation(ts):
    ans = []
    for i in ts:
      
        ans.append((np.random.randn(*i) *0.1 ,np.zeros((i[0] , 1))))

    return ans

params =initialisation([(3 , 2) , (2 , 3) ,(1 , 2)])

for i in params :
    print(f"shape of the weight : {i[0].shape}\nWeight : \n{i[0]}")
    print(f"shape of the bais : {i[1].shape}\nbais : \n{i[1]}")
    print("#"*70)
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
def forward(X , parametres):
    ans = []
    A=X
    for params in parametres:
        Z =params[0] @ A  + params[1]
        A = sigmoid(Z)
        ans.append((Z , A))

    return ans , ans[len(params) - 1][1]

# X = np.random.randn(2 , 1)
X = np.array([[1] , [0]])

his , y_hat = forward(X , params)
print(y_hat)
```

```python
X1 = np.array([0 , 0])
X2 = np.array([0 , 1])
X3 = np.array([1 , 0])
X4 = np.array([1 , 1])
y_hat_list=[]
for i in [X1,X2,X3 , X4]:
    _ , y_hat = forward(i.reshape(-1 , 1) , params)
    print(y_hat)
```
